# 一個電子、兩個氫核
使用原子單位，每個氫核各放一個指數固定為 1 的氫原子 1s 函數。曲線是兩中心模型的總能量，包含核間排斥。演算無隨機步驟。公式推導見 [McCullagh Lab 教材](https://mccullaghlab.github.io/physical_chemistry/quantum_mechanics/Single_Electron_Molecule_Variational_Approach.html)。

```python
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path

def total_energy(distance, sign=1):
    # distance in bohr, energy in hartree; sign = +1 (symmetric) or -1.
    r = np.asarray(distance)
    overlap = np.exp(-r) * (1 + r + r**2 / 3)
    attraction = -1 / r + np.exp(-2 * r) * (1 + 1 / r)
    exchange = -np.exp(-r) * (1 + r)
    return -0.5 + 1 / r + (attraction + sign * exchange) / (1 + sign * overlap)

distance = np.linspace(0.8, 8.0, 361)
bonding = total_energy(distance)
minimum = np.argmin(bonding)
print(f"Grid minimum: R = {distance[minimum]:.2f} bohr; E = {bonding[minimum]:.5f} hartree")
for r in (1., 2., 3., 6.):
    print(f"{r:.1f} bohr: {total_energy(r):.5f} hartree")
assert bonding[minimum] < -0.5
assert abs(total_energy(40.) + 0.5) < 1e-10
```

```python
plt.figure(figsize=(7.6, 4.5), layout="constrained")
plt.axhline(-0.5, color="#565656", linestyle=":", label="Separated H + proton")
plt.plot(distance, bonding, color="#135d72", linewidth=2.6, label="Symmetric combination")
plt.plot(distance, total_energy(distance, -1), color="#b65428", linestyle="--", label="Antisymmetric combination")
plt.scatter(distance[minimum], bonding[minimum], color="#135d72", zorder=3)
plt.xlabel("Distance between nuclei (bohr)")
plt.ylabel("Total energy (hartree)")
plt.legend(frameon=False)
plt.title("One electron shared by two hydrogen nuclei")
folder = Path("../../figures") if Path.cwd().name == "ch05" else Path("figures")
folder.mkdir(parents=True, exist_ok=True)
plt.savefig(folder / "05-01-h2plus.png", dpi=160)
plt.show()
```
