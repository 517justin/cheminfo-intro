# 酚、苯甲醚和環己醇的氧附近有什麼不同？

修改第三格後，重跑第三格與第四格。首次載入 ESOL 需要網路；transformers=[]，溶解度未經標準化。


```python
%%writefile "{script}"
from pathlib import Path
import os
import tempfile
from rdkit import Chem
from rdkit.Chem import Draw, rdDepictor
import deepchem as dc

# ESOL：1,128 筆水溶解度；log10(mol/L)。DOI: 10.1021/ci034243x
# 本節只核對結構和欄位，不訓練模型；保留 scaffold 切分再合併讀取全部資料。
_, parts, _ = dc.molnet.load_delaney(
    featurizer=dc.feat.RawFeaturizer(smiles=True), splitter="scaffold",
    transformers=[], reload=False, data_dir=os.environ.get("DEEPCHEM_DATA_DIR"))
records = []
for part in parts:
    for smiles, y in zip(part.ids, part.y):
        mol = Chem.MolFromSmiles(str(smiles))
        if mol is None:
            raise ValueError(f"ESOL 結構無法讀入：{smiles}")
        records.append((mol, float(y[0])))
# 下面三個字串只作查詢鍵；圖和溶解度皆取自 ESOL 實際資料列。
wanted = [("Phenol", "Oc1ccccc1"), ("Anisole", "COc1ccccc1"),
          ("Cyclohexanol", "OC1CCCCC1")]
selected = []
for name, key in wanted:
    hits = [(m, y) for m, y in records if Chem.MolToSmiles(m) == key]
    assert len(hits) == 1, (name, len(hits))
    mol, value = hits[0]
    selected.append((name, mol, value))
out = Path("../../figures") if Path("../../_quarto.yml").is_file() else Path("figures")
out.mkdir(exist_ok=True)

def oxygen_environment(mol):
    oxygen = next(a for a in mol.GetAtoms() if a.GetSymbol() == "O")
    neighbors = [(a.GetIdx(), a.GetSymbol(), a.GetIsAromatic())
                 for a in oxygen.GetNeighbors()]
    bonds = [str(b.GetBondType()) for b in oxygen.GetBonds()]
    return oxygen.GetTotalNumHs(), neighbors, bonds

for name, mol, _ in selected:
    print(name, "oxygen:", oxygen_environment(mol))
    ring_sizes = [len(r) for r in mol.GetRingInfo().AtomRings()]
    aromatic_atoms = sum(a.GetIsAromatic() for a in mol.GetAtoms())
    aromatic_bonds = sum(b.GetIsAromatic() for b in mol.GetBonds())
    print("  atoms", mol.GetNumAtoms(), "ring sizes", ring_sizes,
          "aromatic atoms/bonds", aromatic_atoms, aromatic_bonds)
    oxygen = next(a for a in mol.GetAtoms() if a.GetSymbol() == "O")
    print("  oxygen in ring", oxygen.IsInRing(), "formal charge", oxygen.GetFormalCharge())
options = Draw.MolDrawOptions()
options.addAtomIndices = True
options.legendFontSize = 23
panel = Draw.MolsToGridImage([m for _, m, _ in selected], molsPerRow=3,
    subImgSize=(350, 270), legends=[name for name, _, _ in selected], drawOptions=options,
    highlightAtomLists=[[a.GetIdx() for a in m.GetAtoms() if a.GetSymbol() == "O"]
                        for _, m, _ in selected])
panel.save(out / "02-02-atoms-bonds-rings.png")
print("ESOL rows:", len(records))
```
